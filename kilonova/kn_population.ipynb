```python
import joblib

import numpy as np
np.random.seed(167834359)

%matplotlib inline
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator, LogLocator
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Computer Modern Roman"],
    "font.size": 12,
    })

import scipy.stats as stats
import scipy.interpolate as interpolate
import pandas as pd

from astropy.cosmology import Planck18
from astropy.time import Time

from nmma.core.conversion import BNSEjectaFitting

# load EOS
m_val, r_val, l_val = np.loadtxt("../eos/RMF3_MRL.dat", unpack=True)
mtov = m_val.max()
r16 = np.interp(1.6, m_val, r_val)
```

```python
def add_kn_parameters(df):
    compactness_1 = df["mass_1_source"] / df["radius_1"] * 1.477
    compactness_2 = df["mass_2_source"] / df["radius_2"] * 1.477
    pc = df["prompt_collapse"] <=1

    # dynamical mass 
    df["log10_mej_dyn"] = np.zeros(df.shape[0])
    df.loc[pc, "log10_mej_dyn"] = np.log10(BNSEjectaFitting().dynamic_mass_fitting_prompt_collapse(mass_1=df["mass_1_source"][pc],
                                                                                              mass_2=df["mass_2_source"][pc],
                                                                                              lambda_1=df["lambda_1"][pc],
                                                                                              lambda_2=df["lambda_2"][pc]))

    df.loc[~pc, "log10_mej_dyn"] = np.log10(BNSEjectaFitting().dynamic_mass_fitting_KrFo(mass_1=df["mass_1_source"][~pc],
                                                                                              mass_2=df["mass_2_source"][~pc],
                                                                                              compactness_1=compactness_1[~pc],
                                                                                              compactness_2=compactness_2[~pc]))
    
    # dynamical velocity
    df["v_ej_dyn"] = np.zeros(df.shape[0])
    df.loc[pc, "v_ej_dyn"] = BNSEjectaFitting().dynamic_vel_fitting_prompt_collapse(mass_1=df["mass_1_source"][pc],
                                                                                  mass_2=df["mass_2_source"][pc],
                                                                                  compactness_1=compactness_1[pc],
                                                                                  compactness_2=compactness_2[pc])
    df.loc[~pc, "v_ej_dyn"] = BNSEjectaFitting().dynamic_vel_fitting_Radice2018(mass_1=df["mass_1_source"][~pc],
                                                                                   mass_2=df["mass_2_source"][~pc],
                                                                                   compactness_1=compactness_1[~pc],
                                                                                   compactness_2=compactness_2[~pc])
    

    # disk mass
    df["log10_mdisk"] = np.zeros(df.shape[0])
    df.loc[pc, "log10_mdisk"] = BNSEjectaFitting().log10_disk_mass_fitting_prompt_collapse(mass_1=df["mass_1_source"][pc],
                                                                                              mass_2=df["mass_2_source"][pc],
                                                                                              lambda_1=df["lambda_1"][pc],
                                                                                              lambda_2=df["lambda_2"][pc])
    df.loc[~pc, "log10_mdisk"] = BNSEjectaFitting().log10_disk_mass_fitting(total_mass=df["mass_1_source"][~pc]+df["mass_2_source"][~pc],
                                                                               mass_ratio=df["mass_2_source"][~pc]/df["mass_1_source"][~pc],
                                                                               MTOV=mtov,
                                                                               R16=r16)
    
    # wind ejecta
    df["log10_mej_wind"] = np.zeros(df.shape[0])
    df.loc[pc, "log10_mej_wind"] = np.log10(stats.truncnorm(loc=0.2, scale=0.1, a=-2, b=2).rvs(size=np.sum(pc))) + df["log10_mdisk"][pc]
    df.loc[~pc, "log10_mej_wind"] = np.log10(stats.truncnorm(loc=0.5, scale=0.2, a=-2.5, b=1.5).rvs(size=np.sum(~pc))) + df["log10_mdisk"][~pc]


    # misc 
    df["v_ej_wind"] = np.random.uniform(0.05, 0.15, size=df.shape[0])
    df["Ye_dyn"] = np.random.uniform(0.15, 0.35, size=df.shape[0])
    df["Ye_wind"] = np.random.uniform(0.2, 0.4, size=df.shape[0])
    df["inclination_EM"] = np.arccos(np.abs(np.cos(df["theta_jn"])))
    
    # for the GRB afterglow
    df["chi_BH"] = BNSEjectaFitting().chiBH_fitting(mass_1=df["mass_1_source"], mass_2=df["mass_2_source"], lambda_1=df["lambda_1"], lambda_2=df["lambda_2"])

    return df
```

## Calculate the light curves

```python
df_narrow = pd.read_csv("full_narrow_kn_parameters.dat", sep=" ")
df_wide = pd.read_csv("full_wide_kn_parameters.dat", sep=" ")
```

```python
from fiesta.inference.lightcurve_model import BullaFlux

model = BullaFlux(name='Bu2026_MLP',
                  filters=["lsstu", "lsstg", "lsstr", "lssti", "lsstz", "lssty"])

def calculate_light_curves(df_in):

    df = df_in.copy()
    df["log10_mej_wind"] = np.maximum(df["log10_mej_wind"], -4)
    df["log10_mej_wind"] = np.minimum(df["log10_mej_wind"], -0.887)

    df["log10_mej_dyn"] = np.maximum(df["log10_mej_dyn"], -4)
    df["log10_mej_dyn"] = np.minimum(df["log10_mej_dyn"], -1.33)

    df["v_ej_dyn"] = np.maximum(df["v_ej_dyn"], 0.12)
    df["v_ej_dyn"] = np.minimum(df["v_ej_dyn"], 0.35)


    kn_parameters = ["inclination_EM", "log10_mej_dyn", "v_ej_dyn", "Ye_dyn", "log10_mej_wind", "v_ej_wind", "Ye_wind", "redshift", "luminosity_distance"]
    params = dict(zip(df[kn_parameters].columns, df[kn_parameters].T.values))
    times, mags = model.vpredict(params)

    return np.array(times), {key: np.array(val) for key, val in mags.items()}



times_wide, mags_wide = calculate_light_curves(df_wide)
times_narrow, mags_narrow = calculate_light_curves(df_narrow)

```

```python
def average_mags(times, mag_array, redshift):
    
    mask = redshift < 1.5
    times_mean = np.geomspace(0.2, 20, 80)

    mag_array = mag_array[mask].copy()
    times = times[mask].copy()

    mags_interp = np.zeros((mag_array.shape[0],times_mean.shape[0]))

    for j in range(mag_array.shape[0]):
        mags_interp[j] = np.interp(times_mean, times[j], mag_array[j])

    mags_median = np.median(mags_interp, axis=0)
    mags_lower = np.quantile(mags_interp, q=0.14, axis=0)
    mags_upper = np.quantile(mags_interp, q=0.86, axis=0)

    return times_mean, mags_median, mags_lower, mags_upper
    
```

```python
def plot_kn_population(ax, df, times, mags, filt="lssti"):
    pc = df["prompt_collapse"] <= 1

    ax[0].hist(df["log10_mej_dyn"][pc], bins=np.linspace(-5, -1, 100), density=False, histtype="step", color="blue", label="prompt collapse fit", rasterized=True)
    ax[0].hist(df["log10_mej_dyn"][~pc], bins=np.linspace(-5, -1, 100), density=False, histtype="step", color="green", label="NS remnant fit", rasterized=True)
    ax[0].vlines([-4, -1.30], *ax[0].get_ylim(), color="lightgrey", linestyle="dashed")
    ax[0].set_xlim(-5, -1)
    ax[0].set_xlabel("$\\log_{10}(m_{\\mathrm{ej, dyn}})$ [$M_\\odot$]", labelpad=0)
    ax[0].set_ylabel("$N$")
    ax[0].set_yscale("log")
    ax[0].set_yticks([1, 10, 100, 1000, 10_000], minor=False)
    
    ax[1].hist(df["log10_mej_wind"][pc], bins=np.linspace(-5, -0, 100), density=False, histtype="step", color="blue", label="prompt collapse fit", rasterized=True)
    ax[1].hist(df["log10_mej_wind"][~pc], bins=np.linspace(-5, -0, 100), density=False, histtype="step", color="green", label="NS remnant fit", rasterized=True)
    ax[1].vlines([-4, -0.55], *ax[1].get_ylim(), color="lightgrey", linestyle="dashed")
    ax[1].set_xlim(-5, 0)
    ax[1].set_xlabel("$\\log_{10}(m_{\\mathrm{ej, wind}})$ [$M_\\odot$]", labelpad=0)
    ax[1].set_ylabel("$N$")
    ax[1].set_yscale('log')
    ax[1].set_yticks([1, 10, 100, 1000, 10_000], minor=False)
    
    ax[2].hist(df["v_ej_dyn"][pc], bins=np.linspace(0.1, 0.5, 100), density=False, histtype="step", color="blue", label="prompt collapse fit", rasterized=True)
    ax[2].hist(df["v_ej_dyn"][~pc], bins=np.linspace(0.1, 0.5, 100), density=False, histtype="step", color="green", label="NS remnant fit", rasterized=True)
    ax[2].vlines([0.12, 0.35], *ax[2].get_ylim(), color="lightgrey", linestyle="dashed")
    ax[2].set_xlim(0.12, 0.35)
    ax[2].set_xlabel("$v_{\\mathrm{ej, dyn}} [c]$", labelpad=0)
    ax[2].set_ylabel("$N$")
    ax[2].set_yscale('log')
    ax[2].set_yticks([1, 10, 100, 1000, 10_000], minor=False)

    times_median, median_pc, upper_pc, lower_pc = average_mags(times[pc], mags[filt][pc], df["redshift"][pc])
    ax[3].plot(times_median, median_pc, color="blue")
    ax[3].plot(times_median, upper_pc, color="blue", linestyle="dashed")
    ax[3].plot(times_median, lower_pc, color="blue", linestyle="dashed")
    inds = np.random.choice(np.arange(mags[filt][pc].shape[0]), replace=False, size=300)
    for ind in inds:
        ax[3].plot(times[pc][ind], mags[filt][pc][ind], color="blue", rasterized=True, alpha=0.05)
    
    times_median, median_nopc, upper_nopc, lower_nopc = average_mags(times[~pc], mags[filt][~pc], df["redshift"][~pc])
    ax[3].plot(times_median, median_nopc, color="green")
    ax[3].plot(times_median, upper_nopc, color="green", linestyle="dashed")
    ax[3].plot(times_median, lower_nopc, color="green", linestyle="dashed")
    inds = np.random.choice(np.arange(mags[filt][~pc].shape[0]), replace=False, size=300)
    for ind in inds:
        ax[3].plot(times[~pc][ind], mags[filt][~pc][ind], color="green", rasterized=True, alpha=0.05)
    
    ax[3].set(xscale="log", xlim=(0.5, 10))
    ax[3].set_xlabel("$t$ [days]", labelpad=0)
    ax[3].set_ylim((32, 23))
    ax[3].set_ylabel(f"apparent mag ${filt[-1]}$")
    ax[3].yaxis.set_major_locator(MaxNLocator(nbins=4))


fig, ax = plt.subplots(4,2, figsize=(8, 9))
fig.subplots_adjust(hspace=0.4, wspace=0.32)

plot_kn_population(ax[:, 0], df_narrow, times_narrow, mags_narrow)
plot_kn_population(ax[:, 1], df_wide, times_wide, mags_wide)

ax[0,0].set_title("Narrow mass distribution", fontsize=15)
ax[0,1].set_title("Wide mass distribution", fontsize=15)

fig.savefig('../figures/kn_population.pdf', dpi=250, bbox_inches="tight")
```
